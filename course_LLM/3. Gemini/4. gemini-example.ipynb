```python
from google import genai
import os
from google.genai import types, errors

client = genai.Client(api_key = os.getenv('GEMINI_API_KEY'))
```

```python
def get_res_of_gemini(
    user_msg:str, messages:list=[],model:str='gemini-3.1-flash-lite'
):
    messages.append(
        types.Content(role='user', parts=[types.Part(text=user_msg)])
    )
    responses = client.models.generate_content(
        model = model,
        contents= messages
    )

    messages.append(
        types.Content(role='model', parts=[types.Part(text=responses.text)])
    )

    return messages
```
